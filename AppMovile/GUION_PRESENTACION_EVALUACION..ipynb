## ❓ ¿Debo explicar código en el video?

**NO.** Las indicaciones y la rúbrica exigen una **demostración práctica y corrida del sistema en ejecución**.
- **No te detengas a leer archivos ni líneas de código:** Hacerlo consumiría minutos valiosos, rompería la fluidez exigida por la rúbrica ("de forma fluida y corrida") y te pondría en riesgo de superar los 10 minutos.
- **Lo que sí debes hacer:** Explicar funcionalmente lo que ocurre en la aplicación mientras señalas simultáneamente en la **consola del IDE** los procesos y logs en tiempo real (peticiones HTTP, tokens, payloads y respuestas del backend).
- **Mención técnica breve:** Al inicio puedes mencionar en 20 segundos la arquitectura usada (*React Native con Expo Router, TypeScript, arquitectura MVVM, Context API para estado global y Axios para consumo de API REST*).

## 🖥️ Configuración Previa de la Grabación (Antes de grabar)

1. **Disposición de la Pantalla (Side-by-Side):**
   - **Mitad Izquierda:** Emulador Android / iOS (o pantalla de tu teléfono físico con scrcpy/mirroring).
   - **Mitad Derecha:** Consola nativa del IDE (VS Code / Antigravity) con la terminal de Expo abierta y visible (`npx expo start`), con letra legible y suficiente espacio vertical para que se aprecien los logs.
2. **Terminal Limpia:**
   - Antes de presionar grabar, limpia la terminal (`clear` o `cls`) para que los logs que aparezcan correspondan exactamente al flujo de la grabación.
3. **Cerrar Sesión Previa:**
   - Asegúrate de que la app inicie en la pantalla de **Login** (`(public)/index.tsx`). Si tenías la sesión guardada, dale a "Log Out" en el Drawer antes de iniciar el video.
4. **Herramienta de Grabación:**
   - Usa **OBS Studio** o la barra de juegos de Windows (`Win + Alt + R`) capturando la pantalla completa a resolución 1080p y con buen audio.

## 🎬 Guión Paso a Paso para la Grabación (Cronometrado ~7 minutos)

### Minuto 0:00 - 0:45 | Introducción y Presentación del Entorno
* **En pantalla:** Emulador a la izquierda en la pantalla de Login; terminal a la derecha limpia y corriendo Expo.
* **Lo que dices:**
  > *"Muy buenos días / tardes, profesor y compañeros. Mi nombre es [Tu Nombre Completo]. A continuación presento la evaluación práctica del segundo parcial de Desarrollo de Aplicaciones Móviles.*  
  > *Como pueden observar en pantalla, a la izquierda tenemos la aplicación móvil desarrollada con React Native, Expo Router y TypeScript ejecutándose en el emulador, y a la derecha tenemos la consola nativa de desarrollo donde monitorearemos los procesos, peticiones HTTP y logs en tiempo real tal como se ha trabajado en clase.*  
  > *Demostraremos los cinco flujos obligatorios: autenticación, actualización de perfil y persistencia, creación de categoría, creación de producto y el filtrado dinámico de productos en la interfaz."*

---

### Minuto 0:45 - 2:00 | Flujo 1: Inicio de Sesión (Autenticación)
* **En pantalla:** En el emulador, introduce tu correo y contraseña registrados.
* **Acción:**
  1. Escribe el correo electrónico: `ejemplo@gmail.com`
  2. Escribe la contraseña. Puedes presionar el ícono del ojo para mostrar/ocultar contraseña (muestra detalle de UI).
  3. Presiona el botón **"Log in"**.
* **Lo que dices:**
  > *"Iniciamos con el flujo de autenticación. Ingreso mis credenciales registradas en la base de datos y presiono 'Log in'.*  
  > *(Señalas la consola derecha)*: Vemos inmediatamente en la consola cómo se envía la petición POST a `/auth/login`, el backend responde satisfactoriamente y recibimos el token Bearer JWT. Este token y los datos del usuario son almacenados de forma persistente mediante nuestro `AuthContext` en `AsyncStorage`, y el sistema nos redirige automáticamente al Dashboard principal protegido."*

---

### Minuto 2:00 - 3:30 | Flujo 2: Actualización de Usuario y Persistencia
* **En pantalla:**
  1. Toca el avatar de perfil en la esquina superior derecha del Dashboard para ir a `/profile`.
  2. Muestra los datos actuales cargados (Nombre, Apellido, Teléfono).
  3. Modifica los campos (por ejemplo, agrega un segundo nombre o cambia el número de teléfono).
  4. Presiona **"Save Changes"**.
  5. Aparecerá la alerta de éxito: *"Tu información ha sido actualizada con éxito en el servidor"*.
  6. Para evidenciar **persistencia**, presiona la flecha de regreso al Dashboard y vuelve a entrar al Perfil para demostrar que los nuevos datos se mantienen intactos.
* **Lo que dices:**
  > *"Ahora nos dirigimos al módulo de gestión de usuario desde el avatar superior. Aquí se precargan los datos de la sesión activa.*  
  > *Procedo a modificar la información, por ejemplo actualizando el nombre y el número de contacto. Al presionar 'Save Changes', observamos en la consola la petición PUT enviada con multipart/form-data hacia el endpoint `/users/upload/:id` incluyendo el token de autorización en los encabezados.*  
  > *El servidor responde exitosamente con estado 200, y la aplicación actualiza tanto el estado global como el almacenamiento local. Si regresamos al Dashboard y volvemos a ingresar a la vista de perfil, comprobamos que los cambios persisten de manera íntegra y no se pierden al recargar la vista."*

---

### Minuto 3:30 - 4:45 | Flujo 3: Creación de Categoría
* **En pantalla:**
  1. En el Dashboard, presiona el ícono del menú hamburguesa (arriba a la izquierda) para desplegar el **Navigation Drawer**.
  2. Despliega la opción **"Inventory"** y selecciona **"Categories"** (te lleva a `inventory.tsx`).
  3. Escribe el nombre de la categoría (ejemplo: `Hogar y Decoración` o `Herramientas`).
  4. Escribe una descripción breve (ejemplo: `Artículos para la renovación y mantenimiento del hogar`).
  5. Presiona el botón **"Guardar Categoria"**.
  6. Se mostrará la alerta: *"!Categoria creada: La categoria '...' se guardó exitosamente"*.
  7. Al presionar "Aceptar", el sistema te devuelve al Dashboard.
* **Lo que dices:**
  > *"Continuamos con la gestión de inventario. Abrimos el menú lateral Drawer, desplegamos 'Inventory' y seleccionamos 'Categories'.*  
  > *Vamos a dar de alta una nueva categoría. Colocamos como nombre 'Hogar y Decoración' y una breve descripción. Al pulsar 'Guardar Categoría', vemos en la consola de la derecha los logs: se imprime el token de sesión utilizado, la llamada POST al endpoint `/categories/create` y la respuesta devuelta por la API con el ID autogenerado y los timestamps de creación. La categoría ha quedado registrada en el sistema."*

---

### Minuto 4:45 - 6:15 | Flujo 4: Creación de Producto
* **En pantalla:**
  1. Abre nuevamente el Drawer lateral, despliega **"Inventory"** y selecciona **"Products"** (te lleva a `create-product.tsx`).
  2. Rellena los campos del formulario:
     - **Nombre del producto:** (ejemplo: `Lámpara LED Inteligente`).
     - **Categoría:** Presiona el selector de categoría. Se abre el Modal dinámico que consulta las categorías disponibles de la API. Selecciona una categoría (ejemplo: *Electronics* o una existente).
     - **SKU / Código de barra:** `PR-902134`.
     - **Descripción detallada:** `Lámpara con control táctil y conectividad WiFi`.
     - **Precio Regular:** `45.50`.
     - **Precio Oferta:** `39.99`.
     - **Stock Inicial:** `25`.
     - **Stock Mínimo:** `5`.
  3. Presiona **"Guardar Producto"**.
  4. Muestra la alerta de confirmación: *"Éxito: Producto creado correctamente"*.
* **Lo que dices:**
  > *"Procedemos a la creación del producto. Ingresamos a la opción 'Products' desde el menú. Como podemos apreciar, la interfaz carga dinámicamente las categorías disponibles mediante el servicio de categorías.*  
  > *Completamos los campos requeridos: nombre del producto, seleccionamos la categoría en el modal interactivo, código SKU, descripción técnica, precio regular, precio en oferta y los niveles de inventario con stock inicial y stock mínimo de alerta.*  
  > *Al presionar 'Guardar Producto', observamos en tiempo real en la consola el payload completo enviado al endpoint `/products/create` mediante el método POST con autenticación Bearer, seguido inmediatamente del log con la respuesta positiva del servidor confirmando que el producto ha sido dado de alta correctamente."*

> 💡 **Nota aclaratoria para ti durante el video:**  
> Tal como indicó el profesor en clase, dado que el backend aún está en proceso de vincular la consulta de productos nuevos por categoría recién creada, el cumplimiento clave de este punto consiste en evidenciar la construcción del payload y los logs en tiempo real de la creación tanto de la categoría como del producto en la consola.

---

### Minuto 6:15 - 7:30 | Flujo 5: Filtrado Dinámico de Productos en la Interfaz
* **En pantalla:**
  1. Estás en el **Dashboard** principal (`dashboard.tsx`).
  2. Muestra la barra horizontal de categorías: `All Categories`, `Electronics`, `Fashions`, `VideoGames`.
  3. Toca **"Electronics"**: La vista se actualiza al instante mostrando únicamente los productos electrónicos (auriculares, cámara).
  4. Toca **"Fashions"**: Se actualiza mostrando los productos de moda (reloj clásico, calzado deportivo).
  5. Toca **"VideoGames"**: Se actualiza mostrando el mando de consola.
  6. Toca **"All Categories"**: Vuelven a mostrarse todos los productos del catálogo.
  7. Opcional (valor agregado): Escribe una palabra en el buscador superior (ejemplo: `"Audio"` o `"Running"`) para mostrar cómo el filtrado dinámico responde tanto por categoría como por búsqueda de texto de forma combinada.
* **Lo que dices:**
  > *"Finalmente, demostramos el último tema visto en clases: el filtrado dinámico de productos por categoría.*  
  > *En nuestro Dashboard contamos con un selector horizontal interactivo implementado con FlatList y un hook `useMemo` reactivo. Si seleccionamos la categoría 'Electronics', la lista filtra inmediatamente mostrando solo dispositivos electrónicos. Al pulsar 'Fashions', la vista renderiza los artículos de moda y calzado; y al elegir 'VideoGames', nos filtra los accesorios de videojuegos. Si regresamos a 'All Categories', se restablece la vista completa.*  
  > *Adicionalmente, este filtrado se sincroniza con el buscador de texto en tiempo real, permitiendo filtrar por título o descripción con alta reactividad y sin recargas de página."*

---

  > *"Con esto completamos de manera corrida los cinco flujos solicitados en la rúbrica de evaluación: inicio de sesión, actualización y persistencia de usuario, registro de categoría, alta de producto con su respectivo monitoreo de logs en consola, y filtrado dinámico en la interfaz de usuario, todo dentro del límite establecido de 10 minutos.*
